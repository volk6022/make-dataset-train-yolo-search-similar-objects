```python
from ultralytics import YOLO
```

```python
model = YOLO("default_models/yolo11x-obb.pt")
```

```python
results = model.train(
    data="datasets/120_annotations/_data.yaml",
    epochs=500,
    patience=70,
    batch=6,
    imgsz=640,
    save=True,
    save_period=20,
    device='cuda:0',
    task='obb',
    warmup_epochs=5,
    close_mosaic=500
)
```
