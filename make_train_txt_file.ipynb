```python
import os

def delete_mismatched_files(image_list_filepath, image_dir="datasets/snd_annotations/images/train", label_dir="datasets/snd_annotations/labels/train"):
    """
    Deletes lines from a text file containing image file paths if the corresponding label file does not exist,
    or if the image file itself doesn't exist.  It assumes a specific file path format for images and labels.

    Args:
        image_list_filepath (str): The path to the text file containing image file paths, one per line in the format:
                                   datasets/snd_annotations/images/train/<filename>.jpg
        image_dir (str): The base directory where images are located.  Defaults to "datasets/snd_annotations/images/train".
        label_dir (str): The base directory where labels are located. Defaults to "datasets/snd_annotations/labels/train".
    """

    existing_lines = []

    try:
        with open(image_list_filepath, 'r', encoding='utf-8') as f:
            for line in f:
                line = line.strip()
                if not line:  # Skip empty lines
                    continue

                # Extract filename from the image path
                try:
                    filename = os.path.basename(line).replace(".jpg", "") # Get filename without extension
                except Exception as e:
                    print(f"Error parsing filename from line: {line} - {e}")
                    continue # skip current line

                # Construct the expected label file path
                label_filepath = os.path.join(label_dir, f"{filename}.txt")

                # Check if both image and label files exist
                # if os.path.exists(line) and os.path.exists(label_filepath):
                if os.path.exists(label_filepath):
                    existing_lines.append(line + '\n')  # Keep the newline
                # else:
                #     if not os.path.exists(line):
                #         print(f"Image file not found, removing from list: {line}")
                #     if not os.path.exists(label_filepath):
                #          print(f"Label file not found, removing associated image from list: {label_filepath} -> {line}")

    except FileNotFoundError:
        print(f"Error: File not found: {image_list_filepath}")
        return

    try:
        print(len(existing_lines))
        with open(image_list_filepath, 'w', encoding='utf-8') as f:
            f.writelines(existing_lines)
        print(f"Successfully updated {image_list_filepath}.")
    except Exception as e:
        print(f"Error writing to file: {e}")


# Example usage:
image_file_list_path = "datasets/snd_annotations/train - Copy.txt"  # Replace with the actual path to your image list file
delete_mismatched_files(image_file_list_path)

# Example with custom directories (optional):
# delete_mismatched_files("image_list.txt", image_dir="my_images", label_dir="my_labels")
```
