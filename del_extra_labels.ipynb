```python
import os
```

```python
def get_image_basenames_from_train_txt(train_txt_path):
    """
    Reads the train.txt file and extracts the base names of the images.
    Example: from 'data/images/image001.jpg' -> 'image001'
    """
    if not os.path.exists(train_txt_path):
        print(f"Error: train.txt file not found at '{train_txt_path}'")
        return None

    valid_image_basenames = set()
    with open(train_txt_path, 'r', encoding="utf-8") as f:
        for line in f:
            image_path = line.strip()
            if image_path: # Ensure line is not empty
                # Get filename (e.g., image001.jpg)
                image_filename = os.path.basename(image_path)
                # Get basename (e.g., image001)
                image_basename, _ = os.path.splitext(image_filename)
                valid_image_basenames.add(image_basename)
    print(f"Found {len(valid_image_basenames)} unique image basenames in '{train_txt_path}'.")
    return valid_image_basenames
```

```python
def clean_redundant_labels(train_txt_path, labels_dir_path, dry_run=True):
    """
    Removes label files from labels_dir_path that do not have a corresponding
    image entry in train_txt_path.
    """
    valid_image_basenames = get_image_basenames_from_train_txt(train_txt_path)
    print(valid_image_basenames)
    if valid_image_basenames is None:
        return

    if not os.path.isdir(labels_dir_path):
        print(f"Error: Labels directory not found at '{labels_dir_path}'")
        return

    redundant_count = 0
    kept_count = 0
    total_labels_checked = 0

    print(f"\nScanning labels directory: '{labels_dir_path}'...")
    for label_filename in os.listdir(labels_dir_path):
        if label_filename.lower().endswith('.txt'): # Process only .txt files
            total_labels_checked += 1
            label_basename, _ = os.path.splitext(label_filename)
            label_filepath = os.path.join(labels_dir_path, label_filename)

            if label_basename not in valid_image_basenames:
                redundant_count += 1
                action = "Would delete" if dry_run else "Deleting"
                # print(f"{action}: '{label_filepath}' (basename '{label_basename}' not in train.txt)")
                if not dry_run:
                    try:
                        os.remove(label_filepath)
                    except OSError as e:
                        print(f"  Error deleting file '{label_filepath}': {e}")
            else:
                kept_count += 1
                # print(f"Kept: '{label_filepath}'") # Optional: for verbose output

    print("\n--- Summary ---")
    print(f"Total label files (.txt) checked in '{labels_dir_path}': {total_labels_checked}")
    print(f"Label files kept: {kept_count}")
    if dry_run:
        print(f"Label files that WOULD BE DELETED (redundant): {redundant_count}")
        if redundant_count > 0:
            print("This was a DRY RUN. No files were actually deleted.")
            print("Run again with --no-dry-run to perform deletions.")
    else:
        print(f"Label files DELETED (redundant): {redundant_count}")
```

```python
train_txt = "datasets/120_annotations/_train.txt"
labels_dir = "datasets/120_annotations/labels/train/"
dry_run=False

print("Starting label cleaning process...")
clean_redundant_labels(train_txt, labels_dir, dry_run)
print("Process finished.")
```
